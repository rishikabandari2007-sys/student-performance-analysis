I performed categorical encoding using three different methods based on the type of categorical data.

1. **Ordinal Encoding:** I used this for categories that have a meaningful order, such as education level, family income, motivation level and exam difficulty. It converts categories into numbers while maintaining their order.

2. **One-Hot Encoding:** I used this for categories that don't have a specific order, such as gender, school type, urban/rural and study method. It creates separate columns for each category with 0s and 1s.

3. **Binary Encoding:** I used this for pass_status because it has only two categories: Pass and Fail. I converted Pass into 1 and Fail into 0.

After encoding, my dataset has 1,00,000 rows and 58 columns. I also checked for missing values, and there are no missing values after encoding.

```python
from scipy.stats import f_oneway
def perform_anova(group_column):

    groups = [
        group['exam_score'].dropna().values
        for name, group in df.groupby(group_column)
    ]

    f_statistic, p_value = f_oneway(*groups)

    print("\nANOVA Analysis:", group_column)
    print("F-Statistic:", f_statistic)
    print("P-Value:", p_value)

    if p_value < 0.05:
        print("Conclusion: Significant difference in average exam scores.")
    else:
        print("Conclusion: No significant difference in average exam scores.")
```

```python
perform_anova('stress_level')
perform_anova('exam_difficulty')
perform_anova('motivation_level')
perform_anova('study_consistency')
```

```python
from sklearn.ensemble import IsolationForest
```

```python
features = [
    'age',
    'previous_exam_score',
    'previous_gpa',
    'attendance_percentage',
    'study_hours_per_day',
    'self_study_hours',
    'online_learning_hours',
    'sleep_hours',
    'daily_screen_time',
    'physical_activity_hours',
    'time_management_score',
    'exam_anxiety_level'
]

X = df[features].copy()

print("Selected features:", X.shape)
print(X.head())
```

```python
iso_forest = IsolationForest(
    contamination=0.05,
    random_state=42
)

iso_forest.fit(X)

print("Isolation Forest model trained successfully!")
```

```python
predictions = iso_forest.predict(X)

df['anomaly_status'] = predictions

print(df['anomaly_status'].value_counts())
```

```python
anomaly_counts = df['anomaly_status'].value_counts()

total = len(df)

print("Total Students:", total)
print("Normal Records:", anomaly_counts[1])
print("Anomalies:", anomaly_counts[-1])

print("Anomaly Percentage:",
      round((anomaly_counts[-1] / total) * 100, 2), "%")
```

```python
anomalies = df[df['anomaly_status'] == -1]

print("Anomalous Student Records:", anomalies.shape)
display(anomalies.head(10))
```

I used Isolation Forest to find unusual patterns in student data.

I selected 12 numerical columns and analysed 1,00,000 student records. The model found 95,000 normal records and 5,000 unusual records (5%).

We did not remove these records because unusual data does not always mean incorrect data. We kept them for further analysis.
